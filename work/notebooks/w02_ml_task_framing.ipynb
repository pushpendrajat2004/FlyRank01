## 1. My lane as an ML task (type)

**Task Type:** Ranking / Scoring.

**Why:** We are trying to prioritize which existing content needs a refresh. Instead of a simple yes/no (classification), we need to know the *relative* importance of updating one page versus another. Scoring allows us to assign a 'decay' or 'opportunity' value to each row, which naturally creates a ranked queue for the content team.

```python
# We can check the distribution of 'trend_direction' to see if it's a good candidate for scoring
df['trend_direction'].value_counts(normalize=True)
```

## 2. Target or proxy

**Target:** The `trend_pct` (percentage change in traffic) over the last period.

**Source:** This is an observed outcome from the traffic logs. By predicting which pages will have the steepest negative `trend_pct`, we identify content that is losing relevance and needs intervention.

```python
# Visualizing the distribution of our target proxy
df['trend_pct'].describe()
```

## 3. Success metric

**Success Metric:** Precision at K (P@K), where K=20.

**Why it matters:** The content team can only refresh a limited number of pages per week. If we provide a list of 20 pages to refresh, we want to maximize the percentage of those pages that were actually in significant decline (e.g., trend_pct < -20%). Success is not just predicting the trend, but ensuring the top of our ranked list is highly actionable.

```python
# Calculate the percentage of 'high urgency' rows currently in the dataset
(df['trend_pct'] < -20).mean() * 100
```

## 4. The unit of analysis, as a real dataframe

**Unit of Analysis:** One row = One unique content page (`content_id`).

**The Slice:** We are looking at all keyword articles across the starter dataset to identify decay patterns. The dataframe below shows the specific grain we are analyzing: performance metrics tied to a unique piece of content.

```python
# Showing the unit of analysis: one row per content_id
lane_slice = df[df['content_type'] == 'keyword article']
lane_slice[['content_id', 'search_volume', 'competition', 'content_age_days', 'trend_pct']].head()
```

## 5. Why ML beats a fixed rule here

**The Messy Pattern:** A fixed rule like "refresh if age > 365 days" is too blunt. Some old pages remain high-performers, while new pages might crash due to high competition or low search intent.

ML can learn the non-linear interactions between `competition_level`, `search_volume`, and `content_age_days`. An if-statement cannot easily weigh how a 10% drop in a high-volume keyword is more critical than a 50% drop in a zero-volume keyword. ML allows us to rank by predicted impact, not just a binary age threshold.

```python
# Demonstrating complexity: Age alone doesn't perfectly correlate with trend
df.groupby('age_tier')['trend_pct'].mean().to_frame()
```
